```python
def test(func):
    def wrap(x):
        print("started")
        func(x)
        print("ended")
    return wrap
```

```python
import time
def timer(func):
    def wrapper(*args,**kwargs):
        start=time.time()
        result=func(*args,**kwargs)
        end=time.time()
        print(f"execution time: {end-start} seconds")
        return result
    return wrapper
@timer
def slow_dp_query():
    time.sleep(1.5)
    return "data fetched"
slow_dp_query()
```

```python
def admin(func):
    def wrapper(user_role,*args,**kwargs):
        if user_role != "admin":
            print("access denied")
            return
        return func(user_role,*args,**kwargs)
    return wrapper
@admin
def delete(admin):
    print("data base deleted")
delete("guest")
delete("admin")
```
